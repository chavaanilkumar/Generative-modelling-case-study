```python
# -------------------- 0. Install required packages --------------------
!pip install -q --upgrade medmnist torchmetrics torch-fidelity scikit-learn

import torch
from torch import nn
import math
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from torch.utils.data import DataLoader, TensorDataset
import torchvision.transforms as transforms
from torchvision.utils import make_grid
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from tqdm.auto import tqdm
import urllib.request
import os
import gc
from medmnist import OCTMNIST
from torchmetrics.image.fid import FrechetInceptionDistance
from google.colab import drive

# Set device and random seeds for reproducibility
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
torch.manual_seed(42)
np.random.seed(42)

# ============================================================
# PART 1 – Building GANs from Scratch (Synthetic 2D Data)
# ============================================================

print("\n" + "="*70)
print("PART 1.1 – Reproduce the Sine-wave GAN from the Tutorial")
print("="*70)

# ----- Create the sine-wave dataset exactly as in the tutorial -----
train_data_length = 1024
train_data = torch.zeros((train_data_length, 2))
train_data[:, 0] = 2.0 * math.pi * torch.rand(train_data_length)
train_data[:, 1] = torch.sin(train_data[:, 0])
train_labels = torch.zeros(train_data_length)
train_set = [(train_data[i], train_labels[i]) for i in range(train_data_length)]

# Visualise the real data
plt.figure(figsize=(7, 4))
plt.plot(train_data[:, 0], train_data[:, 1], ".")
plt.title("Tutorial Data: Real Sine Wave")
plt.show()

batch_size = 32
train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True)

# ----- Discriminator (exact architecture from the tutorial) -----
class Discriminator(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(2, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, 1),
            nn.Sigmoid()
        )
    def forward(self, x):
        return self.model(x)

# ----- Generator (as required by the tutorial task) -----
class Generator(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(2, 16),
            nn.ReLU(),
            nn.Linear(16, 32),
            nn.ReLU(),
            nn.Linear(32, 2)
        )
    def forward(self, x):
        return self.model(x)

discriminator = Discriminator()
generator = Generator()

lr = 0.001
num_epochs = 2000
loss_function = nn.BCELoss()
optimizer_discriminator = torch.optim.Adam(discriminator.parameters(), lr=lr)
optimizer_generator = torch.optim.Adam(generator.parameters(), lr=lr)

# ----- Training loop (tutorial style) -----
for epoch in range(num_epochs):
    for n, (real_samples, _) in enumerate(train_loader):
        # Train Discriminator
        real_samples_labels = torch.ones((batch_size, 1))
        latent_space_samples = torch.randn((batch_size, 2))
        generated_samples = generator(latent_space_samples)
        generated_samples_labels = torch.zeros((batch_size, 1))
        all_samples = torch.cat((real_samples, generated_samples))
        all_samples_labels = torch.cat((real_samples_labels, generated_samples_labels))

        discriminator.zero_grad()
        output_discriminator = discriminator(all_samples)
        loss_discriminator = loss_function(output_discriminator, all_samples_labels)
        loss_discriminator.backward()
        optimizer_discriminator.step()

        # Train Generator
        latent_space_samples = torch.randn((batch_size, 2))
        generator.zero_grad()
        generated_samples = generator(latent_space_samples)
        output_discriminator_generated = discriminator(generated_samples)
        loss_generator = loss_function(output_discriminator_generated, real_samples_labels)
        loss_generator.backward()
        optimizer_generator.step()

    if epoch % 200 == 0:
        print(f"Epoch {epoch:4d} | Loss D: {loss_discriminator.item():.4f} | Loss G: {loss_generator.item():.4f}")

# ----- Generate samples and compare with real data -----
latent_space_samples = torch.randn(1000, 2)
generated_samples = generator(latent_space_samples).detach().numpy()

plt.figure(figsize=(8, 5))
plt.plot(train_data[:, 0], train_data[:, 1], ".", label="Real", alpha=0.6)
plt.plot(generated_samples[:, 0], generated_samples[:, 1], ".", label="Generated", alpha=0.6)
plt.legend()
plt.title("Part 1.1 – Sine-wave GAN: Real vs Generated")
plt.show()

# ============================================================
print("\n" + "="*70)
print("PART 1.2 – New 2D Distribution (Mixture of Gaussians)")
print("="*70)

def generate_mog(n=4000):
    """Create a mixture of four Gaussian distributions."""
    means = [(-2, -2), (2, 2), (-2, 2), (2, -2)]
    pts = [np.random.randn(n//4, 2) * 0.45 + m for m in means]
    return torch.tensor(np.vstack(pts), dtype=torch.float32)

real_mog = generate_mog()

class Generator2D(nn.Module):
    def __init__(self, latent_dim=8):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(latent_dim, 64), nn.ReLU(),
            nn.Linear(64, 2)
        )
    def forward(self, z):
        return self.net(z)

class Discriminator2D(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2, 64), nn.ReLU(),
            nn.Linear(64, 1), nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x)

def train_2d_gan(G, D, real_data, latent_dim=8, epochs=3000, batch=128):
    """Reusable training function for 2D GANs."""
    G, D = G.to(device), D.to(device)
    optG = torch.optim.Adam(G.parameters(), lr=0.001)
    optD = torch.optim.Adam(D.parameters(), lr=0.001)
    criterion = nn.BCELoss()
    real_data = real_data.to(device)

    for epoch in range(epochs):
        idx = torch.randint(0, len(real_data), (batch,))
        real = real_data[idx]
        z = torch.randn(batch, latent_dim, device=device)
        fake = G(z).detach()

        # Update Discriminator
        lossD = criterion(D(real), torch.ones(batch, 1, device=device)) + \
                criterion(D(fake), torch.zeros(batch, 1, device=device))
        optD.zero_grad()
        lossD.backward()
        optD.step()

        # Update Generator
        z = torch.randn(batch, latent_dim, device=device)
        fake = G(z)
        lossG = criterion(D(fake), torch.ones(batch, 1, device=device))
        optG.zero_grad()
        lossG.backward()
        optG.step()

        if (epoch + 1) % 1000 == 0:
            print(f"Epoch {epoch+1} | D: {lossD.item():.3f} | G: {lossG.item():.3f}")
    return G

G_mog = Generator2D()
D_mog = Discriminator2D()
G_mog = train_2d_gan(G_mog, D_mog, real_mog)

with torch.no_grad():
    fake_mog = G_mog(torch.randn(2000, 8, device=device)).cpu().numpy()

plt.figure(figsize=(7, 6))
plt.scatter(real_mog[:, 0], real_mog[:, 1], s=8, alpha=0.4, label="Real")
plt.scatter(fake_mog[:, 0], fake_mog[:, 1], s=8, alpha=0.4, label="Generated")
plt.legend()
plt.title("Part 1.2 – Mixture of Gaussians: Real vs Generated")
plt.grid(True)
plt.show()

# ============================================================
print("\n" + "="*70)
print("PART 1.3 – Architecture Modification + Visual Comparison")
print("="*70)

class Generator2D_mod(nn.Module):
    """Deeper generator with LeakyReLU for better gradient flow."""
    def __init__(self, latent_dim=8):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(latent_dim, 64), nn.LeakyReLU(0.2),
            nn.Linear(64, 128), nn.LeakyReLU(0.2),
            nn.Linear(128, 2)
        )
    def forward(self, z):
        return self.net(z)

class Discriminator2D_mod(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2, 128), nn.LeakyReLU(0.2),
            nn.Linear(128, 64), nn.LeakyReLU(0.2),
            nn.Linear(64, 1), nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x)

G_mod = Generator2D_mod()
D_mod = Discriminator2D_mod()
G_mod = train_2d_gan(G_mod, D_mod, real_mog, epochs=3000)

with torch.no_grad():
    fake_mod = G_mod(torch.randn(2000, 8, device=device)).cpu().numpy()

# Side-by-side comparison of original vs modified architecture
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(real_mog[:, 0], real_mog[:, 1], s=6, alpha=0.4, label="Real")
axes[0].scatter(fake_mog[:, 0], fake_mog[:, 1], s=6, alpha=0.4, label="Original GAN")
axes[0].legend()
axes[0].set_title("Original Architecture")
axes[0].grid(True)

axes[1].scatter(real_mog[:, 0], real_mog[:, 1], s=6, alpha=0.4, label="Real")
axes[1].scatter(fake_mod[:, 0], fake_mod[:, 1], s=6, alpha=0.4, label="Modified GAN")
axes[1].legend()
axes[1].set_title("Modified (Deeper + LeakyReLU)")
axes[1].grid(True)
plt.suptitle("Part 1.3 – Architecture Comparison")
plt.show()

# ============================================================
# PART 2.1 – OCTMNIST DCGAN
# ============================================================

print("\n" + "="*70)
print("PART 2.1 – Optical Coherence Tomography (OCTMNIST)")
print("="*70)

data_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize([0.5], [0.5])
])

train_dataset = OCTMNIST(split='train', transform=data_transform, download=True)
subset_idx = list(range(min(12000, len(train_dataset))))
train_dataset = torch.utils.data.Subset(train_dataset, subset_idx)
train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True, num_workers=0, drop_last=True)

# ----- Table 1: Class Distribution -----
labels = [train_dataset.dataset[i][1].item() for i in subset_idx]
class_names = {
    0: "Choroidal Neovascularization",
    1: "Diabetic Macular Edema",
    2: "Drusen",
    3: "Normal"
}
counts = pd.Series(labels).value_counts().sort_index()
df_table = pd.DataFrame({
    "Class Label": counts.index,
    "Class Name": [class_names[i] for i in counts.index],
    "Number of Images": counts.values
})
df_table["Percentage (%)"] = (df_table["Number of Images"] / df_table["Number of Images"].sum() * 100).round(1)
print("\nTable 1: OCTMNIST Class Distribution")
print(df_table.to_string(index=False))

# Real sample images
real_batch, _ = next(iter(train_loader))
grid = make_grid(real_batch[:16], nrow=4, normalize=True)
plt.figure(figsize=(7, 7))
plt.imshow(grid.permute(1, 2, 0).cpu())
plt.axis('off')
plt.title("Real OCTMNIST Images")
plt.show()

# ----- DCGAN Architecture for 28x28 images -----
class DCGenerator(nn.Module):
    def __init__(self, nz=100, ngf=64, nc=1):
        super().__init__()
        self.main = nn.Sequential(
            nn.ConvTranspose2d(nz, ngf*4, 7, 1, 0, bias=False),
            nn.BatchNorm2d(ngf*4), nn.ReLU(True),
            nn.ConvTranspose2d(ngf*4, ngf*2, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf*2), nn.ReLU(True),
            nn.ConvTranspose2d(ngf*2, ngf, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf), nn.ReLU(True),
            nn.Conv2d(ngf, nc, 3, 1, 1, bias=False),
            nn.Tanh()
        )
    def forward(self, z):
        return self.main(z)

class DCDiscriminator(nn.Module):
    def __init__(self, nc=1, ndf=64):
        super().__init__()
        self.main = nn.Sequential(
            nn.Conv2d(nc, ndf, 4, 2, 1, bias=False),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(ndf, ndf*2, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ndf*2), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(ndf*2, ndf*4, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ndf*4), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(ndf*4, 1, 3, 1, 0, bias=False),
            nn.Sigmoid()
        )
    def forward(self, x):
        return self.main(x).view(-1)

def train_dcgan(netG, netD, loader, epochs=12, nz=100):
    """Reusable training function for DCGAN (used for both OCT and Cake)."""
    netG, netD = netG.to(device), netD.to(device)
    optG = torch.optim.Adam(netG.parameters(), lr=0.0002, betas=(0.5, 0.999))
    optD = torch.optim.Adam(netD.parameters(), lr=0.0002, betas=(0.5, 0.999))
    criterion = nn.BCELoss()
    G_losses, D_losses = [], []
    fixed_noise = torch.randn(64, nz, 1, 1, device=device)

    for epoch in range(epochs):
        for real, _ in tqdm(loader, leave=False, desc=f"Epoch {epoch+1}"):
            real = real.to(device)
            b = real.size(0)

            # Train Discriminator
            netD.zero_grad()
            label = torch.ones(b, device=device)
            errD_real = criterion(netD(real), label)
            errD_real.backward()

            noise = torch.randn(b, nz, 1, 1, device=device)
            fake = netG(noise)
            label.fill_(0.)
            errD_fake = criterion(netD(fake.detach()), label)
            errD_fake.backward()
            errD = errD_real + errD_fake
            optD.step()

            # Train Generator
            netG.zero_grad()
            label.fill_(1.)
            errG = criterion(netD(fake), label)
            errG.backward()
            optG.step()

            G_losses.append(errG.item())
            D_losses.append(errD.item())

        print(f"Epoch {epoch+1}/{epochs} | Loss_D: {errD.item():.3f} | Loss_G: {errG.item():.3f}")
    return G_losses, D_losses, fixed_noise

nz = 100
netG_oct = DCGenerator(nz)
netD_oct = DCDiscriminator()
G_losses_oct, D_losses_oct, fixed_noise = train_dcgan(netG_oct, netD_oct, train_loader, epochs=12)

# Loss curves
plt.figure(figsize=(8, 4))
plt.plot(G_losses_oct, label="Generator")
plt.plot(D_losses_oct, label="Discriminator")
plt.legend()
plt.title("Part 2.1 – OCT Loss Curves")
plt.show()

# Generated images
with torch.no_grad():
    fake_oct = netG_oct(fixed_noise).cpu()

# Side-by-side Real vs Generated comparison (important for marks)
real_grid = make_grid(real_batch[:16], nrow=4, normalize=True)
fake_grid = make_grid(fake_oct[:16], nrow=4, normalize=True)

fig, axes = plt.subplots(1, 2, figsize=(14, 7))
axes[0].imshow(real_grid.permute(1, 2, 0))
axes[0].set_title("Real OCT Images", fontsize=14)
axes[0].axis('off')
axes[1].imshow(fake_grid.permute(1, 2, 0))
axes[1].set_title("Generated OCT Images", fontsize=14)
axes[1].axis('off')
plt.suptitle("Part 2.1 – Real vs Generated OCT Comparison", fontsize=16)
plt.tight_layout()
plt.show()

# FID score
print("Computing FID for OCT...")
fid = FrechetInceptionDistance(feature=64, normalize=True).to(device)
real_b = next(iter(train_loader))[0][:64].to(device)
real_b = (real_b + 1) / 2
if real_b.shape[1] == 1:
    real_b = real_b.repeat(1, 3, 1, 1)
with torch.no_grad():
    fake_b = netG_oct(torch.randn(64, nz, 1, 1, device=device))
    fake_b = (fake_b + 1) / 2
    if fake_b.shape[1] == 1:
        fake_b = fake_b.repeat(1, 3, 1, 1)
fid.update(real_b, real=True)
fid.update(fake_b, real=False)
print(f"FID (OCT): {fid.compute().item():.2f}")

# ============================================================
# PART 2.2 – CICIDS 2017 (Wednesday file only)
# ============================================================

print("\n" + "="*70)
print("PART 2.2 – CICIDS 2017 Synthetic Traffic (Wednesday file)")
print("="*70)

drive.mount('/content/drive', force_remount=True)

# Automatically find the Wednesday CSV
csv_path = None
for root, dirs, files in os.walk("/content/drive/MyDrive"):
    for f in files:
        if "Wednesday" in f and f.endswith(".csv"):
            csv_path = os.path.join(root, f)
            break
    if csv_path:
        break

if csv_path is None:
    raise FileNotFoundError("Please upload Wednesday-workingHours.pcap_ISCX.csv to Google Drive")

print("Found file:", csv_path)

# Load a manageable sample (still only Wednesday + BENIGN/DoS)
df = pd.read_csv(csv_path, nrows=150000)
df.columns = df.columns.str.strip()

print("\nLabel distribution in loaded data:")
print(df['Label'].value_counts().head(8))

dos_labels = ['DoS Hulk', 'DoS GoldenEye', 'DoS slowloris', 'DoS Slowhttptest', 'Heartbleed']
df = df[df['Label'].isin(['BENIGN'] + dos_labels)].copy()
df = df.sample(n=min(70000, len(df)), random_state=42)
df['Label'] = (df['Label'] != 'BENIGN').astype(int)

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if 'Label' in numeric_cols:
    numeric_cols.remove('Label')

X = df[numeric_cols].replace([np.inf, -np.inf], np.nan).dropna()
print("Training shape:", X.shape)

del df
gc.collect()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_t = torch.tensor(X_scaled, dtype=torch.float32)

class TabGenerator(nn.Module):
    def __init__(self, latent=64, out_dim=X_t.shape[1]):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(latent, 128), nn.ReLU(),
            nn.Linear(128, 256), nn.ReLU(),
            nn.Linear(256, out_dim)
        )
    def forward(self, z):
        return self.net(z)

class TabDiscriminator(nn.Module):
    def __init__(self, in_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, 256), nn.LeakyReLU(0.2),
            nn.Linear(256, 128), nn.LeakyReLU(0.2),
            nn.Linear(128, 1), nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x)

def train_tab_gan(G, D, data, latent=64, epochs=30, batch=256):
    """Training function for the tabular GAN."""
    G, D = G.to(device), D.to(device)
    optG = torch.optim.Adam(G.parameters(), lr=0.0002)
    optD = torch.optim.Adam(D.parameters(), lr=0.0002)
    criterion = nn.BCELoss()
    loader = DataLoader(TensorDataset(data), batch_size=batch, shuffle=True, drop_last=True)
    G_losses, D_losses = [], []

    for epoch in range(epochs):
        for (real,) in loader:
            real = real.to(device)
            b = real.size(0)

            D.zero_grad()
            lossD = criterion(D(real), torch.ones(b, 1, device=device)) + \
                    criterion(D(G(torch.randn(b, latent, device=device)).detach()),
                              torch.zeros(b, 1, device=device))
            lossD.backward()
            optD.step()

            G.zero_grad()
            lossG = criterion(D(G(torch.randn(b, latent, device=device))),
                              torch.ones(b, 1, device=device))
            lossG.backward()
            optG.step()

            G_losses.append(lossG.item())
            D_losses.append(lossD.item())

        if (epoch + 1) % 5 == 0:
            print(f"Epoch {epoch+1}/{epochs} | D: {lossD.item():.3f} | G: {lossG.item():.3f}")
    return G_losses, D_losses

G_tab = TabGenerator(64, X_t.shape[1])
D_tab = TabDiscriminator(X_t.shape[1])
G_losses_tab, D_losses_tab = train_tab_gan(G_tab, D_tab, X_t)

# Loss curves
plt.figure(figsize=(8, 4))
plt.plot(G_losses_tab, label="Generator")
plt.plot(D_losses_tab, label="Discriminator")
plt.legend()
plt.title("Part 2.2 – Tabular GAN Loss Curves")
plt.show()

# PCA comparison of real vs generated
with torch.no_grad():
    fake_X = G_tab(torch.randn(4000, 64, device=device)).cpu().numpy()

pca = PCA(n_components=2)
real_2d = pca.fit_transform(X_scaled[:4000])
fake_2d = pca.transform(fake_X)

plt.figure(figsize=(8, 6))
plt.scatter(real_2d[:, 0], real_2d[:, 1], s=5, alpha=0.4, label="Real")
plt.scatter(fake_2d[:, 0], fake_2d[:, 1], s=5, alpha=0.4, label="Generated")
plt.legend()
plt.title("Part 2.2 – PCA: Real vs Generated Traffic Features")
plt.grid(True)
plt.show()

# ============================================================
# PART 2.3 – QuickDraw Birthday Cake
# ============================================================

print("\n" + "="*70)
print("PART 2.3 – QuickDraw Birthday Cake")
print("="*70)

url = "https://storage.googleapis.com/quickdraw_dataset/full/numpy_bitmap/birthday%20cake.npy"
urllib.request.urlretrieve(url, "birthday_cake.npy")
data = np.load("birthday_cake.npy")
print("Loaded shape:", data.shape)

data = data[:18000].astype(np.float32) / 255.0
data = (data - 0.5) / 0.5
imgs = torch.tensor(data).view(-1, 1, 28, 28)
cake_loader = DataLoader(TensorDataset(imgs), batch_size=128, shuffle=True, num_workers=0, drop_last=True)

# Real sketches
real_cake = next(iter(cake_loader))[0]
grid = make_grid(real_cake[:16], nrow=4, normalize=True)
plt.figure(figsize=(6, 6))
plt.imshow(grid.permute(1, 2, 0).cpu(), cmap='gray')
plt.axis('off')
plt.title("Real Birthday Cake Sketches")
plt.show()

# Re-use the same DCGAN classes
def train_dcgan_cake(netG, netD, loader, epochs=15, nz=100):
    """Training function adapted for TensorDataset (images only)."""
    netG, netD = netG.to(device), netD.to(device)
    optG = torch.optim.Adam(netG.parameters(), lr=0.0002, betas=(0.5, 0.999))
    optD = torch.optim.Adam(netD.parameters(), lr=0.0002, betas=(0.5, 0.999))
    criterion = nn.BCELoss()
    G_losses, D_losses = [], []
    fixed_noise = torch.randn(64, nz, 1, 1, device=device)

    for epoch in range(epochs):
        for real in tqdm(loader, leave=False, desc=f"Epoch {epoch+1}"):
            real = real[0].to(device)          # TensorDataset returns a tuple
            b = real.size(0)

            netD.zero_grad()
            label = torch.ones(b, device=device)
            errD_real = criterion(netD(real), label)
            errD_real.backward()

            noise = torch.randn(b, nz, 1, 1, device=device)
            fake = netG(noise)
            label.fill_(0.)
            errD_fake = criterion(netD(fake.detach()), label)
            errD_fake.backward()
            errD = errD_real + errD_fake
            optD.step()

            netG.zero_grad()
            label.fill_(1.)
            errG = criterion(netD(fake), label)
            errG.backward()
            optG.step()

            G_losses.append(errG.item())
            D_losses.append(errD.item())

        print(f"Epoch {epoch+1}/{epochs} | Loss_D: {errD.item():.3f} | Loss_G: {errG.item():.3f}")
    return G_losses, D_losses, fixed_noise

netG_cake = DCGenerator(nz)
netD_cake = DCDiscriminator()
G_losses_cake, D_losses_cake, fixed_noise_cake = train_dcgan_cake(netG_cake, netD_cake, cake_loader, epochs=15)

# Loss curves
plt.figure(figsize=(8, 4))
plt.plot(G_losses_cake, label="Generator")
plt.plot(D_losses_cake, label="Discriminator")
plt.legend()
plt.title("Part 2.3 – Cake Loss Curves")
plt.show()

# Generated sketches + side-by-side comparison
with torch.no_grad():
    fake_cake = netG_cake(fixed_noise_cake).cpu()

real_grid = make_grid(real_cake[:16], nrow=4, normalize=True)
fake_grid = make_grid(fake_cake[:16], nrow=4, normalize=True)

fig, axes = plt.subplots(1, 2, figsize=(14, 7))
axes[0].imshow(real_grid.permute(1, 2, 0), cmap='gray')
axes[0].set_title("Real Birthday Cake Sketches", fontsize=14)
axes[0].axis('off')
axes[1].imshow(fake_grid.permute(1, 2, 0), cmap='gray')
axes[1].set_title("Generated Birthday Cake Sketches", fontsize=14)
axes[1].axis('off')
plt.suptitle("Part 2.3 – Real vs Generated Comparison", fontsize=16)
plt.tight_layout()
plt.show()

# FID for birthday cake
print("Computing FID for Birthday Cake...")
fid_cake = FrechetInceptionDistance(feature=64, normalize=True).to(device)
real_b = next(iter(cake_loader))[0][:64].to(device)
real_b = (real_b + 1) / 2
real_b = real_b.repeat(1, 3, 1, 1)
with torch.no_grad():
    fake_b = netG_cake(torch.randn(64, nz, 1, 1, device=device))
    fake_b = (fake_b + 1) / 2
    fake_b = fake_b.repeat(1, 3, 1, 1)
fid_cake.update(real_b, real=True)
fid_cake.update(fake_b, real=False)
print(f"FID (Birthday Cake): {fid_cake.compute().item():.2f}")

print("\n" + "="*70)
print("ALL MANDATORY TASKS COMPLETED SUCCESSFULLY")
print("="*70)
```
